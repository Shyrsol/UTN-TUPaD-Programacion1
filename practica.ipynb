## Ejercicio 1 — “Caja del Kiosco”

**Objetivo:** Simular una compra con validaciones y cálculo de total.

### Requisitos
1. Pedir nombre del cliente (solo letras, validar con `.isalpha()` en `while`).
2. Pedir cantidad de productos a comprar (número entero positivo, validar con `.isdigit()` en `while`).
3. Por cada producto (usar `for`):
   * Pedir precio (entero, validar `.isdigit()`).
   * Pedir si tiene descuento S/N (validar con `while`, aceptar `s` o `n` en cualquier mayúscula/minúscula).
   * Si tiene descuento: aplicar 10% al precio de ese producto.
4. Al final mostrar:
   * Total sin descuentos
   * Total con descuentos
   * Ahorro total
   * Promedio por producto (usar `float` y formatear con `:.2f`, ejem: `x = 3.14159` -> `print(f"{x:.2f}")`)

### Validaciones obligatorias
* Sin `try/except`.
* No aceptar vacío en nombre (si queda vacío, es error).
* Cantidad > 0 (si ingresa 0, volver a pedir).

### Salida esperada (ejemplo)
```text
Cliente: Ana 
Cantidad de productos: 3 
Producto 1 - Precio: 100  Descuento (S/N): s 
Producto 2 - Precio: 50   Descuento (S/N): n 
Producto 3 - Precio: 200  Descuento (S/N): s 
Total sin descuentos: $350 
Total con descuentos: $320.00 
Ahorro: $30.00 
Promedio por producto: $106.67

````python
# ==========================================
# EJERCICIO 1: CAJA DEL KIOSCO
# ==========================================

# 1. Pedir nombre del cliente (solo letras, no vacío, validar con isalpha en while)
nombre = input("Cliente: ")
while not nombre.isalpha():
    print("Error: El nombre debe contener solo letras y no estar vacío.")
    nombre = input("Cliente: ")
print(f"Cliente: {nombre}")

# 2. Pedir cantidad de productos (número entero positivo > 0, validar con isdigit en while)
cantidad_input = input("Cantidad de productos: ")
while not cantidad_input.isdigit() or int(cantidad_input) <= 0:
    print("Error: Ingrese un número entero positivo mayor a 0.")
    cantidad_input = input("Cantidad de productos: ")

cantidad_productos = int(cantidad_input)
print(f"Cantidad de productos: {cantidad_productos}")

# Variables acumuladoras
total_sin_descuentos = 0.0
total_con_descuentos = 0.0

# 3. Procesar cada producto con for
for i in range(1, cantidad_productos + 1):
    precio_input = input(f"Producto {i} - Precio: ")
    while not precio_input.isdigit() or int(precio_input) <= 0:
        print("Error: El precio debe ser un número entero positivo.")
        precio_input = input(f"Producto {i} - Precio: ")
    
    precio = float(precio_input)
    total_sin_descuentos += precio
    
    tiene_descuento = input(f"Producto {i} - Descuento (S/N): ").lower()
    while tiene_descuento not in ['s', 'n']:
        print("Error: Ingrese 'S' o 'N'.")
        tiene_descuento = input(f"Producto {i} - Descuento (S/N): ").lower()
    
    if tiene_descuento == 's':
        precio_final = precio * 0.90
    else:
        precio_final = precio
        
    total_con_descuentos += precio_final
    
    # Imprimir la línea formateada del producto cargado
    print(f"Producto {i} - Precio: {int(precio)}  Descuento (S/N): {tiene_descuento}")

# 4. Cálculos
ahorro_total = total_sin_descuentos - total_con_descuentos
promedio_por_producto = total_con_descuentos / cantidad_productos

# Salida de totales con salto de línea previo
print()
print(f"Total sin descuentos: ${total_sin_descuentos:.0f}")
print(f"Total con descuentos: ${total_con_descuentos:.2f}")
print(f"Ahorro: ${ahorro_total:.2f}")
print(f"Promedio por producto: ${promedio_por_producto:.2f}")
````

## Ejercicio 2 — “Acceso al Campus y Menú Seguro”

**Objetivo:** Sistema de login con control de intentos y menú interactivo repetitivo.

### Requisitos
1. Definir credenciales fijas: `usuario = "alumno"`, `clave = "python123"`.
2. Permitir un máximo de 3 intentos para validar credenciales.
3. Si falla los 3 intentos: mostrar `"Cuenta bloqueada"` y terminar la ejecución.
4. Si ingresa correctamente: mostrar un menú repetitivo (`while`) hasta seleccionar salir:
   1. Ver estado de inscripción (mostrar `"Inscripto"`)
   2. Cambiar clave (pedir nueva clave de mínimo 6 caracteres y confirmación; deben coincidir)
   3. Mostrar mensaje motivacional
   4. Salir
5. Validaciones:
   * Menú: entrada numérica (`.isdigit()`) entre 1 y 4.
   * Cambio de clave: la nueva clave debe tener al menos 6 caracteres (`len() >= 6`).
   * Sin uso de `try/except` ni listas.

````python
# ==========================================
# EJERCICIO 2: ACCESO AL CAMPUS
# ==========================================

# Parámetros del sistema
USUARIO_CORRECTO = "alumno"
intentos = 3
autenticado = False

# 1. Autenticación (Máximo 3 intentos)
while intentos > 0 and not autenticado:
    usuario = input("Usuario: ")
    clave = input("Contraseña: ")
    
    # Validación de credenciales
    # Clave debe tener >= 6 caracteres y ser distinta del usuario
    if usuario == USUARIO_CORRECTO and len(clave) >= 6 and clave != usuario:
        autenticado = True
        print("\n¡Acceso concedido al Campus Virtual!")
    else:
        intentos -= 1
        if intentos > 0:
            print(f"Credenciales incorrectas. Intentos restantes: {intentos}\n")
        else:
            print("\nAcceso bloqueado. Ha superado el límite de 3 intentos.")

# 2. Menú de opciones (Solo si inició sesión)
if autenticado:
    opcion = ""
    while opcion != "4":
        print("\n--- MENÚ DEL CAMPUS ---")
        print("1. Ver materias inscriptas")
        print("2. Cargar nota de examen")
        print("3. Consultar estado académico")
        print("4. Salir")
        
        opcion = input("Seleccione una opción (1-4): ")
        
        # Validar que sea un dígito entre 1 y 4
        while not opcion.isdigit() or int(opcion) < 1 or int(opcion) > 4:
            print("Error: Opción inválida. Debe ingresar un número entre 1 y 4.")
            opcion = input("Seleccione una opción (1-4): ")
            
        if opcion == "1":
            print(">> Materias inscriptas: Programación I, Base de Datos, Matemática.")
        elif opcion == "2":
            print(">> Nota cargada correctamente en el sistema.")
        elif opcion == "3":
            print(">> Estado académico: Regular / En cursada.")
        elif opcion == "4":
            print(">> Sesión finalizada. Hasta luego.")
````

## Ejercicio 3 — “Agenda de Turnos”

**Objetivo:** Simular la asignación de turnos médicos en un rango de horario determinado, procesando las solicitudes de los pacientes y validando las entradas.

### Requisitos
1. Pedir la cantidad de turnos disponibles para el día (número entero positivo > 0, validar con `while` y `.isdigit()`).
2. Por cada turno disponible (usar ciclo `for`):
   * Pedir el nombre del paciente (solo letras y espacios, no vacío, validar con `while`).
   * Pedir la hora del turno (entero entre 8 y 18 hs, validar con `while` y `.isdigit()`).
   * Mostrar el turno asignado en formato de línea individual.
3. Al finalizar la carga de todos los turnos, mostrar un resumen con:
   * Total de turnos asignados.
   * Primer turno del día (hora).
   * Último turno del día (hora).
4. **Validaciones obligatorias:**
   * Sin `try/except`.
   * Sin listas (usar variables para seguir el primer y último turno).

````python
# ==========================================
# EJERCICIO 3: AGENDA DE TURNOS
# ==========================================

# 1. Pedir cantidad de turnos disponibles (entero > 0, validar con isdigit)
turnos_input = input("Cantidad de turnos disponibles: ")
while not turnos_input.isdigit() or int(turnos_input) <= 0:
    print("Error: Ingrese un número entero positivo mayor a 0.")
    turnos_input = input("Cantidad de turnos disponibles: ")

total_turnos = int(turnos_input)
print(f"Cantidad de turnos disponibles: {total_turnos}")

# Variables de seguimiento (sin listas)
primer_hora = 0
ultima_hora = 0

# 2. Procesar cada turno con ciclo for
for i in range(1, total_turnos + 1):
    
    # Validar nombre del paciente (solo letras y espacios, no vacío)
    paciente = input(f"Paciente {i} - Nombre: ")
    while not paciente.replace(" ", "").isalpha() or len(paciente.strip()) == 0:
        print("Error: El nombre debe contener solo letras y no estar vacío.")
        paciente = input(f"Paciente {i} - Nombre: ")
    
    # Validar hora del turno (entero entre 8 y 18 hs)
    hora_input = input(f"Paciente {i} - Hora (8 a 18): ")
    while not hora_input.isdigit() or int(hora_input) < 8 or int(hora_input) > 18:
        print("Error: La hora debe ser un número entero entre 8 y 18.")
        hora_input = input(f"Paciente {i} - Hora (8 a 18): ")
    
    hora = int(hora_input)
    print(f"Paciente {i} - Nombre: {paciente}  Hora (8 a 18): {hora}")
    
    # Registro de límites de horario
    if i == 1:
        primer_hora = hora
    ultima_hora = hora

# 3. Resumen final
print()
print(f"Total de turnos asignados: {total_turnos}")
print(f"Primer turno del día: {primer_hora}:00 hs")
print(f"Último turno del día: {ultima_hora}:00 hs")
````

## Ejercicio 4 — “Escape Room: La Bóveda”

**Objetivo:** Desarrollar un programa en Python que gestione una misión de escape bajo restricciones de energía, tiempo y seguridad, validando acciones y recursos sin usar listas ni `try/except`.

### Requisitos
1. **Variables de Estado iniciales:**
   * Energía (ej: 100)
   * Tiempo (ej: 30 minutos)
   * Cerraduras abiertas (contador de 0 a 3)
   * Alarma / Seguridad (control de acciones consecutivas o riesgo)
2. **Ciclo de juego para abrir las 3 cerraduras:**
   * Mostrar opciones de acción: 
     1. Forzar cerradura (consume mucha energía, bajo tiempo)
     2. Hackear panel (consume tiempo medio, riesgo de alarma)
     3. Descansar (recupera energía, consume tiempo)
   * Validar la entrada de la opción con `while` y `.isdigit()`.
3. **Restricciones y Condiciones:**
   * Controlar que la energía y el tiempo no lleguen a cero o valores negativos (derrota).
   * Si se activa la condición de alarma, la misión falla.
   * Si se abren las 3 cerraduras dentro de los recursos, la misión es exitosa.

````python
# ==========================================
# EJERCICIO 5: CAJERO AUTOMÁTICO SIMPLIFICADO
# ==========================================

saldo = 50000
opcion = ""

print("--- CAJERO AUTOMÁTICO ---")

while opcion != "4":
    print("\n1) Consultar saldo")
    print("2) Depositar dinero")
    print("3) Retirar dinero")
    print("4) Salir")
    
    opcion = input("Opción (1-4): ")
    
    while not opcion.isdigit() or int(opcion) < 1 or int(opcion) > 4:
        print("Error: Ingrese un número válido entre 1 y 4.")
        opcion = input("Opción (1-4): ")
        
    if opcion == "1":
        print(f"Saldo actual: ${saldo}")
        
    elif opcion == "2":
        monto_input = input("Monto a depositar: ")
        while not monto_input.isdigit() or int(monto_input) <= 0:
            print("Error: El monto debe ser mayor a 0.")
            monto_input = input("Monto a depositar: ")
        
        deposito = int(monto_input)
        saldo += deposito
        print(f"Depósito exitoso. Nuevo saldo: ${saldo}")
        
    elif opcion == "3":
        monto_input = input("Monto a retirar: ")
        while not monto_input.isdigit() or int(monto_input) <= 0:
            print("Error: El monto debe ser mayor a 0.")
            monto_input = input("Monto a retirar: ")
            
        retiro = int(monto_input)
        
        while retiro > saldo:
            print(f"Error: Fondos insuficientes. Saldo disponible: ${saldo}")
            monto_input = input("Monto a retirar: ")
            while not monto_input.isdigit() or int(monto_input) <= 0:
                print("Error: El monto debe ser mayor a 0.")
                monto_input = input("Monto a retirar: ")
            retiro = int(monto_input)
            
        saldo -= retiro
        print(f"Retiro exitoso. Nuevo saldo: ${saldo}")
        
    elif opcion == "4":
        print("Gracias por utilizar el cajero automático. Saliendo...")
````

## Ejercicio 5 — “Escape Room: La Arena del Gladiador”

**Objetivo:** Desarrollar un programa en Python que gestione un combate simulado en una arena bajo restricciones de salud y rondas, validando acciones sin usar listas ni `try/except`.

### Requisitos
1. **Variables de Estado iniciales:**
   * Salud del gladiador (ej: `100`)
   * Victorias requeridas para escapar (ej: `3`)
2. **Ciclo de combate (hasta ganar 3 rondas o perder la salud):**
   * Mostrar opciones de combate:
     1. Ataque rápido
     2. Ataque pesado
     3. Defender y curar
   * Validar la entrada de la opción con `while` y `.isdigit()`.
3. **Restricciones obligatorias:**
   * Sin `try/except`.
   * Sin listas.

````python
# ==========================================
# EJERCICIO 5: ESCAPE ROOM - LA ARENA DEL GLADIADOR
# ==========================================

salud = 100
victorias = 0

print("--- INICIO DE PRUEBA: LA ARENA DEL GLADIADOR ---")
print("Objetivo: Sobrevivir y derrotar a 3 oponentes para escapar.")

while victorias < 3 and salud > 0:
    print(f"\nEstado actual -> Salud: {salud} | Victorias: {victorias}/3")
    print("1) Ataque rápido (-10 Salud, +1 Victoria)")
    print("2) Ataque pesado (-25 Salud, +1 Victoria con fuerza bruta)")
    print("3) Defender (+15 Salud, sin avance)")
    
    opcion = input("Opción (1) Ataque rápido  2) Ataque pesado  3) Defender): ")
    
    while not opcion.isdigit() or int(opcion) < 1 or int(opcion) > 3:
        print("Error: Ingrese un número válido entre 1 y 3.")
        opcion = input("Opción (1) Ataque rápido  2) Ataque pesado  3) Defender): ")
        
    accion = int(opcion)
    print(f"Acción elegida: {accion}")
    
    if accion == 1:
        salud -= 10
        victorias += 1
        print("¡Ataque rápido exitoso! Oponente derrotado.")
        
    elif accion == 2:
        salud -= 25
        victorias += 1
        print("¡Golpe crítico! Oponente eliminado con fuerza bruta.")
        
    elif accion == 3:
        salud += 15
        if salud > 100:
            salud = 100
        print("Te has defendido y recuperado salud.")

# Evaluación del resultado final
print("\n--- RESULTADO DE LA ARENA ---")
if victorias >= 3:
    print("¡Victoria! Has vencido a todos los oponentes y escapado de la arena.")
else:
    print("Derrota: Tu salud ha llegado a cero en la arena.")
````
